### Assigning Avg. Age to the null values in Age column

```python
# this will iterate one by one and if its na it will return avg are or else the normal value of x
def assign_age(x):
    if pd.isna(x):
        return Avg_Age
    return x
```

```python
data["Age"]=data["Age"].apply(assign_age)
```

```python
df["Age"]=df["Age"].fillna(Avg_Age)
```
