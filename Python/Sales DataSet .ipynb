```python
df1=df["category"].value_counts().reset_index()
```

```python
sns.set_palette("Paired")
```

```python

bars=sns.barplot(data=df1,x="category",y="count")
plt.bar_label(bars.containers[0],padding=3)
plt.title("Category Wise Sales")
```

```python
df2=df.groupby("subcategory")["profit"].sum().sort_values(ascending=True).reset_index()
df2
```

```python
bars=plt.barh(df2["subcategory"],width=df2["profit"])
plt.bar_label(bars,padding=1)
plt.title("Most Profitable Subcategory")
plt.xlabel("Profits")
plt.ylabel("Subcategory")
```

```python
df_state_sale=df.groupby("state")["sales"].sum().sort_values(ascending=False).head(10).reset_index()
```

```python
sns.barplot(data=df_state_sale,   x=df_state_sale["state"],   y=df_state_sale["sales"])
plt.xticks(rotation=45)
plt.title("Most Profitable States")
```

```python
sns.set_theme(style="whitegrid")
sns.palette="flare"
```

```python
df3=df.groupby("region")["sales"].sum()
```

```python
plt.pie(df3,autopct='%1.1f%%',labels=df3.index)
plt.legend()
plt.title("Region Wise Sales")
```

```python
sns.scatterplot(data=df,x="sales",y="profit",hue="category")
plt.title("Sales And Profit")
```

```python
Ship_mode=df["shipmode"].value_counts().reset_index()
Ship_mode
```

```python
bars=sns.barplot(data=Ship_mode,x="shipmode",y="count")
plt.bar_label(bars.containers[0],padding=2)
plt.title("Most Common Shipping Mode")
```

```python
Month=df.groupby("orderdate_month")["sales"].sum().reset_index()
Month
```

```python
plt.plot(Month["orderdate_month"],Month["sales"],marker="^",markerfacecolor="red")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Per Month")
```
