## Top Goal Scorers by Nationality

```python
df1 = (df.groupby("nationality")["total_goals_tournament"].sum().sort_values(ascending=False).head(10).reset_index())

plt.figure(figsize=(10,6))
sns.barplot(data=df1,x="nationality",y="total_goals_tournament",palette="viridis")

plt.xticks(rotation=45)
plt.title("Top 10 Nationalities by Total Goals")
plt.xlabel("Nationality")
plt.ylabel("Goals")
```

```python
df1 = ( df.groupby("team")["tournament_rating"].mean().sort_values(ascending=False).head(15).reset_index())

plt.figure(figsize=(10,7))

sns.barplot(data=df1,x="tournament_rating",y="team",palette="mako")

plt.title("Top Teams by Average Tournament Rating")
plt.xlabel("Average Rating")
plt.ylabel("Team")
```

## Average Goals by Position

```python
df1 = ( df.groupby("position")["total_goals_tournament"] .mean() .sort_values(ascending=False) .reset_index())

plt.figure(figsize=(8,5))

sns.barplot(data=df1,x="position",y="total_goals_tournament", palette="rocket"
)

plt.title("Average Goals by Position")
plt.xlabel("Position")
plt.ylabel("Average Goals")
```

## Average Rating by Position

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="position",
    y="tournament_rating",
    palette="crest"
)

plt.title("Tournament Rating by Position")
plt.xlabel("Position")
plt.ylabel("Tournament Rating")
```
